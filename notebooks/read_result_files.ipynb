```python
import pandas as pd
from pathlib import Path    
```

```python
result_path = Path("../results/")
methods = ["logistic_regression", "mutual_information", "random", "random_forest", "uniform"]
datasets = ["breast_cancer", "folktables_employment", "folktables_income", "hr_analytics", "loan_prediction"]
bias_types = ["less_negative_class", "less_positive_class", "mean_difference"]
metrics = ["MMDs", ]
```

```python
data = []
for dataset in datasets:
    for bias_type in bias_types:
        for method in methods:
            json = pd.read_json(str(result_path / method / "downstream" / dataset / bias_type / "results.json"))
            mmd = [dataset, bias_type, method, json["MMDs"]["mean"]]
            data.append(mmd)
```

```python
data_frame = pd.DataFrame(data=data, columns=["Dataset", "Bias Type", "Method", "Mean MMD"])
```
