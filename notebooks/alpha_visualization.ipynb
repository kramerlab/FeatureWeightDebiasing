```python
import numpy as np
import seaborn as sns
sns.set_style("darkgrid")
import matplotlib.pyplot as plt
from pathlib import Path

Path("diagrams").mkdir(exist_ok=True)
```

```python
prediction = np.linspace(0, 1, 1000, endpoint=False)
difference = 0.5 - prediction
k = 0.1
weighting = 1.5 - (1 / (1 + np.exp(difference / k)))
```

```python
ax = sns.lineplot(x=prediction, y= weighting)
ax.set(xlabel='Probability', ylabel='Weight')
x_ticks = [x / 10 for x in range(0, 11)]
_ = ax.set_xticks(x_ticks)

y_ticks = [y / 10 for y in range(5, 16)]
_ = ax.set_yticks(y_ticks)
plt.savefig("diagrams/exponential_soft_mrs_weights.pdf")
```

```python
weighting = 1.5 - prediction
ax = sns.lineplot(x=prediction, y= weighting)
ax.set(xlabel='Probability', ylabel='Weight')
x_ticks = [x / 10 for x in range(0, 11)]
_ = ax.set_xticks(x_ticks)

y_ticks = [y / 10 for y in range(5, 16)]
_ = ax.set_yticks(y_ticks)
plt.savefig("diagrams/linear_soft_mrs_weights.pdf")
```
