```python
import seaborn as sns
import json
```

```python
dataset = "breast_cancer"
bias_type = "mean_difference"
path = f"../results/downstream/{dataset}/{bias_type}/repeated_MRS/uniform_feature_weighting/"
with open(path + "mmd_with_feature_weights.json") as file:
    mmd_with_feature_weights = json.load(file)

with open(path + "mmd_without_feature_weights.json") as file:
    mmd_without_feature_weights = json.load(file)
```

```python
sns.lineplot(mmd_with_feature_weights[:], label="With Feature Weights")
ax = sns.lineplot(mmd_without_feature_weights[:], label="Without Feature Weights")
ax.set(title=f'MRS on {dataset} with 50% Budget')
_ = ax.set(xlabel='Number of Dropped Elements', ylabel='Weighted MMD')
```

```python
with open(path + "mutual_information_sum.json") as file:
    mutual_information_list = json.load(file)
```

```python
ax = sns.lineplot(mutual_information_list[:])
ax.set(title=f'MRS on {dataset}')
_ = ax.set(xlabel='Number of Dropped Elements', ylabel='Mutual Information Sum over Features')
```
