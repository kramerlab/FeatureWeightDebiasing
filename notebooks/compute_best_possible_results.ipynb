```python
import numpy as np
from pathlib import Path
import json
```

```python
result_path = Path("../results/downstream_task")
bias_type = "none"
less_bias_strength = "0.0"
method = "uniform"
data_set_replacer = {
    "diabetes": "Diabetes",
    "folktables_employment": "Employment",
    "folktables_income": "Income",
    "bank_marketing": "Bank Marketing",
    "hr_analytics": "HR Analytic",
    "german_credit": "German Credit",
    "breast_cancer": "Breast Cancer",
    "loan_prediction": "Loan",
    }
```

```python
for data_set_name in data_set_replacer.keys():
    auroc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_auroc_list.json"
    auprc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_auprc_list.json"

    pad_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_pad_list.json"
    domain_auroc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_domain_auroc_list.json"
    try:
        with open(auroc_file_name) as file:
            auroc_list = json.load(file)
            print(f"AUROC: {data_set_name}: ${np.round(np.mean(auroc_list), 3)} \\pm {np.round(np.std(auroc_list), 2)}$")

        with open(auprc_file_name) as file:
            auprc_list = json.load(file)
            print(f"AUPRC: {data_set_name}: ${np.round(np.mean(auprc_list), 3)} \\pm {np.round(np.std(auprc_list), 2)}$ \n")


        with open(pad_file_name) as file:
            pad_list = json.load(file)
            print(f"MMD: {data_set_name}: ${np.round(np.mean(pad_list), 3)} \\pm {np.round(np.std(pad_list), 2)}$ \n")

    except FileNotFoundError:
        continue
```
