```python
import numpy as np
from pathlib import Path
import json
```

```python
result_path = Path("../results/downstream_task")
bias_type = "none"
less_bias_strength = "0.0"
method = "uniform"
data_set_replacer = {
    "diabetes": "Diabetes",
    "folktables_employment": "Employment",
    "folktables_income": "Income",
    "bank_marketing": "Bank Marketing",
    "hr_analytics": "HR Analytic",
    "german_credit": "German Credit",
    "breast_cancer": "Breast Cancer",
    "loan_prediction": "Loan",
    }

data_set_targets = {"diabetes": "Diabetes_binary_relative_bias",
                    "folktables_employment": "Employment_relative_bias", 
                    "folktables_income": "Binary Income_relative_bias",
                    "bank_marketing": "y_relative_bias",
                    "hr_analytics": "target_relative_bias",
                    "german_credit": "class_relative_bias", 
                    "breast_cancer": "class_relative_bias", 
                    "loan_prediction": "Loan_Status_relative_bias",
                    }
```

```python
for data_set_name in data_set_replacer.keys():
    auroc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_auroc_list.json"
    auprc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_auprc_list.json"
    similarity_T_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "similarity_results_T.json"

    with open(auroc_file_name) as file:
        auroc_list = json.load(file)
        print(f"AUROC: {data_set_name}: ${np.round(np.mean(auroc_list), 3)}\\pm{np.round(np.std(auroc_list), 2)}$")

    with open(auprc_file_name) as file:
        auprc_list = json.load(file)
        print(f"AUPRC: {data_set_name}: ${np.round(np.mean(auprc_list), 3)}\\pm{np.round(np.std(auprc_list), 2)}$")

    with open(similarity_T_file_name) as file:
        similarity_T_dict = json.load(file)
        mmd_T_mean = similarity_T_dict["MMDs"]["mean"]
        mmd_T_std = similarity_T_dict["MMDs"]["sd"]
        print(f"MMD T: {data_set_name}: ${np.round(mmd_T_mean, 4)}\\pm{np.round(mmd_T_std, 3)}$")

        relative_bias_T_mean = similarity_T_dict[data_set_targets[data_set_name]]["bias mean"]
        relative_bias_T_std = similarity_T_dict[data_set_targets[data_set_name]]["bias sd"]
        print(f"Relative Bias T: {data_set_name}: ${np.round(relative_bias_T_mean, 2)}\\pm{np.round(relative_bias_T_std, 2)}$ \n")
```
