```python
from ucimlrepo import fetch_ucirepo
import numpy as np
```

```python
cdc_diabetes_health_indicators = fetch_ucirepo(id=891) 
```

```python
cdc_diabetes_health_indicators
diabetes_X = cdc_diabetes_health_indicators.data.features 
diabetes_y = cdc_diabetes_health_indicators.data.targets 
```

```python
print(len(diabetes_y))
print(np.sum(diabetes_y == 0))
print(np.sum(diabetes_y == 1))
print(len(diabetes_X.columns))
```

```python
diabetes_X.to_csv("diabetes_X.csv", index=False)
diabetes_y.to_csv("diabetes_y.csv", index=False)
```
