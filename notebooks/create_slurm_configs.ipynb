```python
methods_list = ["uniform","psa", "kmm", "mrs-forest", "soft-mrs-exponential", "fw-mrs-temperature",  "fw-mrs-temperature-svm"]
bias_types = ["less_negative_class", "less_positive_class", "mean_difference"]
less_bias_strengths = "0.1"
mean_bias_strengthts = "0.8"
datasets = ["folktables_employment", "folktables_income", "hr_analytics", "breast_cancer", "loan_prediction", "german_credit",
    "bank_marketing", "diabetes"]
```

```python
print("ArrayTaskID BiasFraction BiasType Method Dataset Drop")
i= 1
for data_set in datasets:
    for bias_type in bias_types:
        bias_strength = mean_bias_strengthts if bias_type == "mean_difference" else less_bias_strengths
        for method in methods_list:
            drop = 1 if data_set in ("breast_cancer", "loan_prediction") else 5
            print(f"{i} {bias_strength} {bias_type} {method} {data_set} {drop}")
            i += 1
```
