```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from src.utils.data_loader import load_brast_cancer_data
import numpy as np
```

```python
df, columns, target = load_brast_cancer_data()
feature_weights = [1, 2, 3, 4, 5, 6, 7, 8, 9]
```

```python
from src.utils.sampling import sample

N, R = sample(
            "mean_difference",
            df,
            target,
            train_fraction=0.5,
            bias_fraction=0.75,
            columns=columns,
        )
```

```python
np.random.seed(5)
tree_clf = DecisionTreeClassifier(splitter="best", max_features="sqrt", random_state=5)
# forest_clf = RandomForestClassifier(splitter="feature_weighted", max_features="sqrt", n_estimators=5, random_state=0)
clf = tree_clf.fit(N[columns], N[target], feature_weights=feature_weights, switch_sample_side=True)
clf.score(R[columns], R[target])
```
