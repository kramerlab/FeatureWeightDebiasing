```python
import numpy as np
from pathlib import Path
import json
```

```python
result_path = Path("../results/downstream_task")
bias_type = "less_positive_class"
less_bias_strength = "0.1"
method = "uniform"
data_set_replacer = {
                    "diabetes", 
                    "folktables_employment", 
                    "folktables_income",
                    "bank_marketing",
                    "hr_analytics", 
                    "german_credit", 
                    "breast_cancer", 
                    "loan_prediction",
                    }
```

```python
for data_set_name in data_set_replacer:
    auroc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_auroc_list.json"
    domain_auroc_file_name = result_path / data_set_name / bias_type / less_bias_strength / method / "classification_results" / "R_domain_auroc_list.json"
    try:
        with open(auroc_file_name) as file:
            auroc_list = json.load(file)
            print(f"AUROC: {data_set_name}: & ${np.round(np.mean(auroc_list), 3)}\\pm{np.round(np.std(auroc_list), 2)}$")

        with open(domain_auroc_file_name) as file:
            domain_auroc_list = json.load(file)
            print(f"Domain AUROC: {data_set_name}:  ${np.round(np.mean(domain_auroc_list), 3)}\\pm{np.round(np.std(domain_auroc_list), 2)}$ \n")

    except FileNotFoundError:
        continue
```
