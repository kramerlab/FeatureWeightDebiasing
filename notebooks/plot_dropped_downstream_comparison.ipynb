```python
import seaborn as sns
import numpy as np
rng = np.random.default_rng(5)
import pandas as pd
sns.set_style("whitegrid")
```

```python
budgets = ["None", 0.5, 0.1, 0.05, 0.01, 0.005, 0.001]
dropped_samples_dict = {}
auroc_dict = {}
data_list = []
for i, budget in enumerate(budgets):
    dropped_samples_dict[budget] =   rng.normal(500 - 10 * i, 10, size=1000)
    auroc_dict = {"Temperature": budget, "Values": rng.normal(0.7, 0.05, size=1000)}
    data_list.append(dropped_samples_dict)
    data_list.append(auroc_dict)
data = pd.DataFrame(data_list)
```

```python
data = data.explode("Values")
```

```python
ax = sns.boxplot(data=dropped_samples_dict)
# ax.set_ylim(0)
```

```python
ax = sns.boxplot(data=data[data["Type"] == "AUROC"], x="Temperature", y="Values")
ax.set_ylim(0.5)
ax.set_ylabel("AUROC")
```

```python
ax = sns.boxplot(data=data[data["Type"] == "Dropped Samples"], x="Temperature", y="Values")
ax.set_ylabel("Dropped Samples")
```
