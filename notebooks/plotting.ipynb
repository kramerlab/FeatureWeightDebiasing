```python
import seaborn as sns
import json
```

```python
with open("../mmd_with_feature_weights.json") as file:
    mmd_with_feature_weights = json.load(file)

with open("../mmd_without_feature_weights.json") as file:
    mmd_without_feature_weights = json.load(file)
```

```python
sns.lineplot(mmd_with_feature_weights, label="With Feature Weights")
ax = sns.lineplot(mmd_without_feature_weights, label="Without Feature Weights")
ax.set(title='MRS on Breast Cancer')
ax.set(xlabel='Number of Dropped Elements', ylabel='Weighted MMD')
```
